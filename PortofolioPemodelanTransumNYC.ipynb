```python
import pandas as pd
from sodapy import Socrata

# Use your real App Token if you have it, otherwise use None
client = Socrata("data.ny.gov", None)

# In this dataset, the column for subway data is often 'transit_mode' 
# but let's remove the 'where' filter for a second just to see what columns exist.
results = client.get(
    "jsu2-fbtj",
    limit=50000,
)

df = pd.DataFrame.from_records(results)
print("Available Columns:", df.columns) # This will tell us the real names
print(df.head())
```

```python
transition_matrix = df.pivot_table(
    index='origin_station_complex_name', 
    columns='destination_station_complex_name', 
    values='transition_probability',
    fill_value=0 
)

print(transition_matrix.shape) 
```

```python
# Check the top 10 destinations from a major hub
hub_name = "Times Sq-42 St" # Or use Penn Sta-34 St
hub_data = df[df['origin_station_complex_name'] == hub_name]
print(hub_data[['destination_station_complex_name', 'estimated_average_ridership']].head(10))
```
