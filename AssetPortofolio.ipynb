```python
%pip install yfinance

import yfinance as yf
import seaborn as sns
import matplotlib.pyplot as plt

tickers = ['BTC-USD', 'NVDA', 'GLD', 'TLT', 'SPY', 'ETH-USD']
prices = yf.download(tickers, start='2020-01-01', auto_adjust=True)['Close']
returns = prices.pct_change(fill_method=None).dropna()

corr_matrix = returns.corr()

plt.figure(figsize=(10,8))
sns.heatmap(corr_matrix,
            annot=True,
            fmt=".2f",
            cmap='RdBu_r',
            center=0,
            vmin=-1, vmax=1,
            linewidths=.5)


plt.title('asset corelation matrix')
plt.show()
```

```python
import yfinance as yf
import numpy as np

tickers= ['BTC-USD', 'NVDA', 'GLD']
prices = yf.download(tickers, start='2020-01-01', auto_adjust=True)['Close']
returns = prices.pct_change(fill_method= None).dropna()

weights = np.rray([0.3,0.3,0.4])

cov_matrix = returns.cov().values *252

port_variancce = np.dot(weights.T, np.dot(cov_matrix, weights))
port_volatility = np.sqrt(port_variancce)

avg_daily_returns = returns.mean()
annualized_returns = avg_daily_returns * 252
port_expected_return = np.dot(weights, annualized_returns)

print(f"Potofolio Volatility: {port_volatility:.4f}")
print(f"Portofolio Expected Return: {port_expected_return:.4f}")
```
